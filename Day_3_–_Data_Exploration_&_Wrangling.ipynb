```python
#dropna(), fillna()
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df=pd.read_csv("employees_missing.csv")
print(df)
drop=df.dropna(axis=0)
print(drop)
#fillna=df[["Salary","Age"]].fillna(df[["Salary","Age"]].mean())
fillna=df.fillna({"Salary":df["Salary"].mean(),"Age":df["Age"].median(),"Experience":df["Experience"].mean(),"City":df["City"].mode(),"Rating":df["Rating"].min()})
print(fillna)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df=pd.read_csv("students.csv")
df1=pd.read_csv("grades.csv")
print(df)
print(df1)
# MERGE
inner_merge=pd.merge(df,df1, on="StudentID")
print("INNER MERGE  :\n",inner_merge)
left_merge=pd.merge(df,df1,on="StudentID",how="left")
print("LEFT MERGE   :\n",left_merge)
right_merge=pd.merge(df,df1,on="StudentID",how="right")
print("RIGHT MERGE   :\n",right_merge)
outer_merge=pd.merge(df,df1,on="StudentID",how="outer")
print("OUTER MERGE   :\n",outer_merge)
# CONCATENATION
column_concat=pd.concat([df,df1],axis=1)
print("Column concatenation: \n",column_concat)
#Join
df=df.set_index("StudentID")
df1=df1.set_index("StudentID")
join=df.join(df1)
print("combines data frames  :\n",join)
join=df.join(df1,how="inner")
print(join)
join=df.join(df1,how="left")
print(join)
join=df.join(df1,how="right")
print(join)
join=df.join(df1,how="outer")
print(join)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
customers=pd.read_csv("customers.csv")
orders=pd.read_csv("orders.csv")
print(customers)
print(orders)
customers_orders=pd.merge(customers,orders,on="CustomerID")
print(customers_orders)
result=customers_orders.groupby("CustomerID")["Amount"].sum()
print(result)
result=customers_orders.groupby("Name")["Amount"].sum()
print(result)
result=customers_orders.groupby("City")["Amount"].sum()
print(result)
result=customers_orders.groupby("Product")["Amount"].sum()
print(result)
result=customers_orders.groupby("CustomerID")["Amount"].agg(Total_amount="sum",Average_amount="mean",number_of_orders="count")
print(result)
result=customers_orders.groupby(["City","Product"])["Amount"].agg(Total_amt="sum")
print("Group by city and product\n",result)
#result=customers_orders.groupby("column_name")["numeric_column"].sum()
result=customers_orders.groupby("CustomerID")["Amount"].agg(Total_amt="sum",Avg_amt="mean",no_of_orders="count",Minimum_amt="min",Maximum_amt="max")
print(result)
result=customers_orders.groupby("City")["Amount"].agg("sum")
print("Total amount by city   :",result)
result=customers_orders.groupby("Product")["Amount"].agg(["sum"])
print(result)
result=customers_orders.groupby("City")["Amount"].agg(["sum","mean","min","max","count"])
print(result)
result=customers_orders.groupby("CustomerID")["OrderID"].count()
print(result)
result=customers_orders.groupby("City").agg(Total_amt=("Amount","sum"),Order_count=("OrderID","count"),Average_amt=("Amount","mean"))
print(result)
```
