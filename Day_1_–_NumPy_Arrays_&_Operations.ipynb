```python
import numpy as np
```

```python
# creating !D,2D and 3D array
#1D array
a=np.array([40,50,45,60,70,65])
#2D array
b=np.array([[40,50,35],[60,70,55]])
#3D array
c=np.array([[[30,40,20],[60,55,45]],[[50,40,65],[80,30,50]],[[60,40,50],[50,60,70]]])

print("One Dimensional array   :  \n ",a)
print("Two Dimensional array   :  \n ",b)
print("Three Dimensional array :  \n ",c)
```

```python
#addition, subtraction, multiplication, division.
a=np.array([20,40,30,80])
b=np.array([2,4,6,8])
print(a+b)
print(a-b)
print(a*b)
print(a/b)
```

```python
#use function sum(),mean(),std()
a=np.array([40,50,45,60,70,65])
print(f"Mean               :{np.mean(a)}")
print(f"Sum                :{np.sum(a)}")
print(f"Standard deviation :{round(np.std(a),2)}")
```

```python
#use function reshape()
a=np.arange(1,13)
print(a)
b=a.reshape(2,6)#changing the shape of array
print(b)
```

```python
#arange(),linspace()
a=np.arange(2,11,2)#specify the steps size
b=np.linspace(2,11,4)#specify the number of values

print("arange  :",a)
print("linspace:",b)
```

```python
# max(),min()
a=np.array([40,50,45,60,70,65])
print(f"Maximum value in this array  :{np.max(a)}")
print(f"Minimum value in this array  :{np.min(a)}")
print(f"Index of maximum             :{np.argmax(a)}")
print(f"Index of minimum             :{np.argmin(a)}")
```
