```python
from matplotlib.lines import lineStyles
from matplotlib import axes
#Day 5 – Matplotlib Visualization
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df=pd.read_csv("students_scores.csv")
print(df)
score_hours=df.groupby("StudyHours")["Score"].mean().reset_index()
print(score_hours)
#Line chart
plt.figure(figsize=(8,6))
plt.plot(score_hours.StudyHours,score_hours.Score)
plt.title("Score by hours")
plt.xlabel("scores")
plt.ylabel("Study hours")
plt.plot(score_hours["StudyHours"],score_hours["Score"],marker="s",color="red",
         linestyle="--",
         linewidth="2",markersize=5)
plt.show()
#scatter plot
plt.scatter(score_hours.StudyHours,score_hours.Score,color="red",
       edgecolors="yellow",alpha=.5,marker="o")
plt.title("Score by hours")
plt.xlabel("StudyHours")
plt.ylabel("Scores")
plt.grid(color="green",linestyle=":",axis="both")
plt.tight_layout()
plt.show()
# bar chart
pc_sales=pd.read_csv("/content/product_category_sales.csv")
print(pc_sales)
plt.bar(pc_sales.Category,pc_sales.Revenue,color="yellow",edgecolor="red",
        alpha=.8,linewidth=2,align="center")
plt.title("Category by sales")
plt.xlabel("Category")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
#pie chart
b_share=pd.read_csv("/content/browser_share.csv")
print(b_share)
browser_share=b_share.groupby("Browser")["SharePct"].mean().reset_index()
plt.figure(figsize=(6,4))
plt.pie(browser_share.SharePct,labels=browser_share.Browser,
        colors=["skyblue","red","blue","green","yellow"],autopct="%1.1f%%",
        explode=(0,0,0,0.2,0),shadow=True)
plt.title("Sharing Browser")
plt.show()
```
