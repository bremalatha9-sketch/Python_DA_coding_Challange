```python
#Day 7– Data Cleaning & Stats
import pandas as pd
import numpy as np
df=pd.read_csv("/content/Titanic dataset.csv")
print(f"Before cleaning  :\n,{df}")
print(f"information about dataset  :\n{df.info()}")
print(f" Null information   :\n{df.isnull().sum()}")
```

```python
df=df.drop_duplicates()
print(df)
```

```python
df["Age"]=df["Age"].fillna(df["Age"].median())

print(df["Age"])
df["Cabin"]=df["Cabin"].fillna("unknown")
df["Embarked"]=df["Embarked"].fillna(df["Embarked"].mode()[0])
df["Embarked"].isnull().sum()
print(df.head())
```

```python
#Step 4: Correct inconsistent data
# Remove extra spaces and standardize text
df["Name"]=df["Name"].str.strip()
df["Sex"]=df["Sex"].str.strip().str.lower()
df["Ticket"]=df["Ticket"].str.strip().str.upper()
df["Embarked"]=df["Embarked"].str.strip().str.upper()
print(df.head())
```

```python
print("Mean:", df["Age"].mean())
print("Median:", df["Age"].median())
print("Mode:", df["Age"].mode()[0])
print("Standard Deviation:", df["Age"].std())
print("Variance:", df["Age"].var())
print(df.isnull().sum())
```

```python
#groupby
print(f"Average age by gender    :\n{df.groupby("Sex")["Age"].mean()}")
print(f"Number of passengers by passenger class   :\n {df.groupby("Pclass")["PassengerId"].mean()}")
print(f"Survival count by gender      :\n{df.groupby("Sex")["Survived"].sum()}")
print(f"Survival percentage by passenger class      :\n{df.groupby("Pclass")["Survived"].mean()*100}")
```

```python
#Final check
print(df.isnull().sum())
print("Dataset shape:", df.shape)
```
