```python
#Day 6 – Seaborn & Plotly Advanced Visualization

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("/content/iris (1).csv")
print(df)
corr=df.corr(numeric_only=True)
sns.heatmap(corr,annot=True,cmap="coolwarm")
plt.show()
```

```python
sns.pairplot(df,vars=["sepal_length","petal_length","sepal_width","sepal_length"],hue="species")
plt.show()
```

```python
sns.violinplot(x="species",y="sepal_length",data=df)
plt.show()
```

```python
sns.swarmplot(df,x="species",y="sepal_length")
plt.show()
```

```python
sns.kdeplot(df,x="sepal_length",fill=True)
plt.title("Length of density")
plt.xlabel("Length")
plt.ylabel("Density")
plt.show()
```

```python
#Interactive Plotly charts (scatter, bar, line).
import plotly.express as px
fig=px.scatter(df,x="sepal_length",y="petal_length")
fig.show()
```

```python
fig=px.bar(df,x="species",y="sepal_length")
fig.show()
```

```python
fig=px.line(df,x="species",y="sepal_length")
fig.show()
```
